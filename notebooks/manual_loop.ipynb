# Manual loop

Drive physics yourself: set a control, step, render. The panel's run/pause only applies to `run()`.

```python
import time

from mujoco_viser_usage import ViserParser

robot = ViserParser(port=8081)
robot.set_control("shoulder_pan", -1.3)
```

```python
for _ in range(250):  # 5 s of simulation at 2 ms steps
    robot.step(nstep=10)
    robot.render()
    time.sleep(0.02)
```

```python
q = robot.get_q()
position, rotation = robot.get_body_pose("wrist_3_link")
print(q.round(3), position.round(3))
```

```python
robot.close()
```
